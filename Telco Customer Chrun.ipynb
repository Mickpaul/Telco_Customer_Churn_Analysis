```python
#df['TotalCharge'] Selects the TotalCharges column.  
#.replace(old value, new value)
# .replace(' ', np.nan)
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)                                   
```

```python
#now pandas recongize it a missing value NaN instead of blank 
df.isnull().sum()
```

```python
#check the datatype
df['TotalCharges'].dtype
```

```python
#the pd.to_numeric() function to convert the datatype of that column into integer or float since we already know its a float
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
```

```python
#Verifiy the the Totalcharge has now been converted numerical value 
df['TotalCharges'].dtypes
```

```python
#use .dropna() to drop or remove any row/column with NaN, since its just 11 missing value, its really insignificant
#(inplace = True) modifies the dataframe, instead of creating a new dataframe 
df.dropna(inplace=True)
#verifiy again that rows/columns with NaN values have been removed 
df.isnull().sum()
```

```python
#Recheck for Duplicates again
df.duplicated().sum()
```

```python
#create a loop that iterate through each columns and check the unique values
#for column in the dataframe columns, iterate and print the each unique values in the each columns in the dataframe
for column in df.columns:
    print(column)
    print(df[column].unique())
    print("-----------------")
    
```
